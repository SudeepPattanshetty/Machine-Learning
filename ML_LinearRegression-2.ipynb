## Visualizing sample data:

```python
x = [10, 20, 5, 100, 50, 18, 100, 60, 30] # Bills = Independent Variable
y = [0, 0, 1, 5, 5, 2, 10, 5, 0] # Tips = Dependent Variable

plt.figure(figsize=(10, 5))
plt.grid(True)
plt.xlabel("Food Bill")
plt.ylabel("Tips given")
plt.plot(x, y, 'o', color='blue')
for xi,yj in zip(x, y):
    label = f"({xi}, {yj})"

    plt.annotate(label, # this is the text
                 (xi,yj), # this is the point to label
                 textcoords="offset points", # how to position the text
                 xytext=(0,10), # distance from text to points(x,y)
                 ha='center') # horizontal alignment can be left, right or center
```

## Understanding linear function

```python
x2 = range(0, 100, 10)
y2 = []

m = 1 # slope
b = 0 # constant
# Formula to find y [ Y = mX + b ]
for xi in x2:
    y_temp = (m*xi) + b
    y2.append(y_temp)

plt.figure(figsize=(10,5))
plt.grid(True)
plt.plot(x2, y2, '-*', color='red')

for xi, yj in zip(x2,y2):
    label = f"({xi}, {yj})"
    plt.annotate(label,
                 (xi, yj),
                 textcoords="offset points",
                 xytext=(0,10),
                 ha="center")
```

## Fitting line to the data

```python
plt.figure(figsize=(10,5))
plt.grid(True)
plt.plot(x, y, 'o', color='blue')
for xi,yj in zip(x, y):
    label = f"({xi}, {yj})"
    plt.annotate(label,
                (xi, yj),
                textcoords="offset points",
                xytext=(0,10),
                ha="center")

x_lr = []
y_lr = []
m=0.1
b=-1.5
for xi in x:
    x_lr.append(xi)
    output_y=(m * xi) + b
    y_lr.append(output_y)

plt.xlabel("Food Bill")
plt.ylabel("Tips given")
plt.plot(x_lr, y_lr, '-', color='red')
```

## Root Mean Squared Error

```python
d = [1, 0.5, 1, 1, 0.5, 1, -0.5, -1, -2]
print(d)
```

```python
sd = []
for di in d:
    sd.append(di * di)

print(sd)
```

```python
sum = 0
for sdi in sd:
    sum += sdi

print(sum)
```

```python
import math
rmse = math.sqrt(sum)
print(rmse)
```
